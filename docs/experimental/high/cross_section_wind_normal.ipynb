# Experimental: cross-section of the wind component normal to the section, with a map

**This is an experimental example.** It reproduces the Metview gallery example [Cross Section Normal Wind Component with Map](https://metview.readthedocs.io/en/latest/gen_files/gallery/cross_section_wind_normal.html) with earthkit-geo, xarray and earthkit-plots. It uses the new `earthkit.geo.polyline` module, whose interface may still change. The figure has two panels: a map with the 500 hPa geopotential height (contours) and the cross-section line (red) above, and the cross-section of the wind normal component below.

The normal wind component is the wind across the section, i.e. the wind component perpendicular to the line. It is positive towards the left of the direction of travel, which is north-west here, because the line goes from south-west to north-east. It is computed by projecting the wind on the unit normal vector of the line.

This is the simplified version of the [raw notebook](../raw/cross_section_wind_normal.ipynb). It uses the `cross_section()` and `wind_components()` helpers from [../xsection.py](../xsection.py) instead of the hand-written regridding, projection and Xarray steps.

The notebook reads a GRIB file. It contains forecast data of superstorm Sandy on 2012-10-29 00 UTC.

To make this notebook work [earthkit-data](https://earthkit-data.readthedocs.io), [earthkit-plots](https://earthkit-plots.readthedocs.io) and xarray have to be installed.

## Imports

```python
import sys
from datetime import datetime

import earthkit.data as ekd
import earthkit.plots as ekp
import numpy as np
from earthkit.geo.polyline import sample_polyline

sys.path.insert(0, "..")  # xsection.py is in docs/experimental
from xsection import cross_section, wind_components
```

## Load the data

The file contains the fields `r`, `u`, `v` and `z` on the pressure levels 1000, 925, 850, 700, 500, 400, 300, 250, 200, 150 and 100 hPa, for a single date and a single step (the analysis time, step 0). It is a GRIB1 file on a regular 0.25° latitude-longitude grid over the area 25–40°N, 84–60°W. The wind components `u` and `v` are the east and north components, because the file is not relative to the grid. Here is a short listing of the first fields.

```python
ds = ekd.from_source("sample", "sandy_pl_025.grib").to_fieldlist()
print(len(ds), "fields")
ds[:6].ls()
```

## Select the wind components and the geopotential at 500 hPa

We select `u` and `v` on all 11 levels, and `z` on 500 hPa only. The geopotential is in m²/s². Dividing it by 98.0665 m²/s² per dam (standard gravity 9.80665 m/s² times 10 m) converts it to geopotential height in dam, which is what Metview contours. The `latitude` and `longitude` coordinates of the field are kept.

```python
fl = ds.sel({"parameter.variable": ["u", "v"]})
z500 = ds.sel({"parameter.variable": "z", "vertical.level": 500})
print(len(fl), "u and v fields")
print(len(z500), "z500 field")

# geopotential (m2/s2) -> geopotential height (dam)
z500_dam = z500.to_xarray()["z"] / 98.0665
z500_dam.attrs = {"units": "dam", "long_name": "Geopotential height"}
```

## Define the line

[sample_polyline()](https://earthkit-geo.readthedocs.io) returns the latitudes and longitudes of 200 points along the line from (30.30°N, 79.83°W) to (36.95°N, 63.92°W), which is the line of the Metview example. We use `method="linear"`: the points are evenly spaced in latitude and longitude, i.e. they lie on a straight line in a plain latitude-longitude map, as in Metview. This is not the great circle, which is the default.

```python
lats, lons = sample_polyline([30.30, 36.95], [-79.83, -63.92], n=200, method="linear")
print(lats.shape, (lats[0], lons[0]), (lats[-1], lons[-1]))
```

## Build the cross-section Dataset with the helper

The `cross_section()` helper from `../xsection.py` does the regridding and builds the Xarray Dataset in one call. The fieldlist is converted to Xarray with `to_xarray()`, and the data is regridded onto the 200 points of the line with `earthkit.geo.regrid()`. The helper uses nearest-neighbour interpolation by default. The result is an Xarray Dataset with the wind components `u` and `v` as variables on `level` (hPa) and `distance` (km, cumulative along the line). The latitudes and longitudes of the points are stored as auxiliary coordinates of `distance`, and the levels are sorted in ascending order. The result follows the CF conventions, and the forecast time is kept as scalar coordinates (`forecast_reference_time`, `step` and `valid_time`).

During the regridding earthkit may log an `Error setting gridSpec` message with a traceback (a `GeocalculusError`), and MIR may print some `ECCODES` lines about the grid. Both are harmless: the regridding is not affected and the result is correct.

```python
xs = cross_section(fl, lats, lons)
xs
```

## Compute the normal wind component

The `wind_components()` helper adds the wind components normal to and parallel with the line to the Dataset, as the variables `wn` and `wp`. At each level and point the normal component is the scalar product of the wind vector with the unit normal vector of the line, `wn = u * east + v * north`, where `east` and `north` are the components of the normal vector returned by [polyline_normals()](https://earthkit-geo.readthedocs.io). The normal vector is on the left of the direction of travel. Here the line goes towards the north-east, so the normal vectors point towards the north-west: `east` is negative and `north` is positive. The normal component is positive towards the left of the direction of travel, i.e. towards the north-west.

The `method` must be the same as the one used in `sample_polyline()`. The helper returns a copy of the Dataset, so `u` and `v` are unchanged. It also adds `wp`, which we do not use here (it is shown in the [parallel notebook](cross_section_wind_parallel.ipynb)).

```python
xs = wind_components(xs, method="linear")
wn = xs["wn"]
print(wn.dims, "NaNs:", int(wn.isnull().sum()))
print(f"wn from {wn.min().item():.1f} to {wn.max().item():.1f} m/s")
```

**Sign convention.** The sign of the normal component matches the Metview plot. At 850 hPa the component is negative (about -21 m/s on average) on the western half of the line and positive (about +20 m/s) on the eastern half. In Metview the western part of the section is orange (negative) and the eastern part is purple (positive). The line crosses the cyclone, which has a counter-clockwise circulation: on its western side the wind blows towards the south, i.e. against the north-west normal, and on its eastern side it blows towards the north, i.e. along the normal. So `wind_components()` has the same orientation as the normal vector of Metview, positive towards the left of the direction of travel. The next cell shows the mean values at 850 hPa.

```python
wn850 = xs["wn"].sel(level=850)
half = xs.sizes["distance"] // 2
print(f"mean wn at 850 hPa, western half: {wn850[:half].mean().item():.1f} m/s")
print(f"mean wn at 850 hPa, eastern half: {wn850[half:].mean().item():.1f} m/s")
```

## Plot the map and the cross-section in one figure

We use one earthkit-plots `Figure` with two rows: a map in the first row and an ordinary (non-map) subplot in the second row, which is taller. The panels are added with `add_map()` and `add_subplot()`.

**Map.** The area is 25–40°N, 84–60°W, in a plain latitude-longitude projection. It has grey land, light blue sea, coastlines and 5° grid lines. The geopotential height at 500 hPa is drawn with black contour lines every 4 dam, with labels. The cross-section line is red and thick, and the title is "Z 500 hPa".

**Cross-section.** The filled contours of `wn` use the levels from -36 to 36 m/s every 4 m/s, with the matplotlib colormap `PuOr`. It goes from orange for negative values to purple for positive values. This is the same palette as the ColorBrewer `PuOr` reversed, because ColorBrewer's own order goes from purple to orange. The colour bar is added with the `legend()` method of the figure. The vertical axis is the logarithmic pressure, with 1000 hPa at the bottom and 100 hPa at the top. The horizontal axis has 8 ticks where the line crosses the longitudes 78°W, 76°W, … and 64°W. They are labelled with the latitude (2 decimals) and the longitude of the crossing, which are found by linear interpolation along the line.

Two details need explanation:

- `resample=False` is needed for `contourf()` of the cross-section. By default earthkit-plots resamples the data onto a 1000 × 1000 latitude-longitude grid, which is wrong for a distance-pressure section.
- The logarithmic axis, the axis limits, the ticks and the labels are set with matplotlib calls on the axis `.ax` of the subplot, because earthkit-plots has no dedicated option for them.

```python
import cartopy.crs as ccrs  # plain latitude-longitude projection for the map

base_date = datetime(2012, 10, 29, 0)
wn_levels = list(range(-36, 37, 4))

lats, lons, distance = xs["latitude"].values, xs["longitude"].values, xs["distance"].values

fig = ekp.Figure(rows=2, columns=1, figsize=(9, 12), height_ratios=[3, 4])

# --- map: z500 and the cross-section line ---
m = fig.add_map(domain=[-84, -60, 25, 40], crs=ccrs.PlateCarree())  # [west, east, south, north]
m.land(color="#888888")
m.ocean(color="#c6d1e4")
m.coastlines()
m.gridlines(xstep=5, ystep=5, color="#434343", linewidth=0.8)
m.contour(
    z500_dam,
    style=ekp.styles.Contour(levels=list(range(400, 640, 4)), colors="black", labels=True, linewidths=1.2),
    resample=False,
)
m.line(x=[lons[0], lons[-1]], y=[lats[0], lats[-1]], color="red", linewidth=3)
m.title("Z 500 hPa")

# --- cross-section: wind normal component ---
c = fig.add_subplot()
c.contourf(
    xs["wn"],
    x="distance",
    y="level",
    style=ekp.styles.Style(levels=wn_levels, colors="PuOr"),
    resample=False,
)
fig.legend(subplots=[c])

# 8 ticks where the line crosses the longitudes 78W ... 64W (lons increase along the line)
tick_lons = np.arange(-78, -63, 2)
tick_distance = np.interp(tick_lons, lons, distance)
tick_lats = np.interp(tick_lons, lons, lats)
tick_labels = [f"{lat:.2f}°N/{abs(lon)}°W" for lat, lon in zip(tick_lats, tick_lons)]

ax = c.ax
ax.set_yscale("log")
ax.set_ylim(1000, 100)  # 1000 hPa at the bottom, 100 hPa at the top
pressure_ticks = [1000, 925, 850, 700, 500, 400, 300, 250, 200, 150, 100]
ax.set_yticks(pressure_ticks)
ax.set_yticklabels([str(p) for p in pressure_ticks])
ax.minorticks_off()
ax.set_xlim(distance[0], distance[-1])
ax.set_xticks(tick_distance)
ax.set_xticklabels(tick_labels)
ax.set_ylabel("Pressure (hPa)")
ax.set_title(f"Cross section of wind normal component – {base_date:%Y-%m-%d %H} UTC, step 0")

fig.show()
```
