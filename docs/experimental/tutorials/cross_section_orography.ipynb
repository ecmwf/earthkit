# Cross-section with orography

A vertical cross-section shows how a quantity changes with height along a line over the Earth's surface. In this tutorial we plot the temperature along a line from (40°N, 10°W) to (56°N, 10°E) and show the ground in black, using the surface pressure.

This feature is **experimental**. For a more detailed version of this example, see [the high-level notebook](../high/cross_section_orography.ipynb).

## Imports

First we import the tools we need.

```python
import sys
from datetime import datetime, timedelta

import earthkit.data as ekd
import earthkit.plots as ekp
import numpy as np
from earthkit.geo.polyline import sample_polyline

sys.path.insert(0, "..")  # xsection.py is in docs/experimental
from xsection import cross_section
```

## Get the data

We load a small sample file and pick the temperature (`t`) and the surface pressure (`sp`) for one date and forecast step.

```python
ds = ekd.from_source("sample", "xs_pl_input.grib1").to_fieldlist()

base_date = datetime(2011, 12, 15, 12)
step = timedelta(hours=12)

fl = ds.sel({"parameter.variable": ["t", "sp"], "time.base_datetime": base_date, "time.step": step})
```

## Define the line

We choose 200 evenly spaced points along the line from (40°N, 10°W) to (56°N, 10°E).

```python
lats, lons = sample_polyline([40, 56], [-10, 10], n=200)
```

## Compute the cross-section

Now we compute the values of the data at the points of the line. Log messages may appear while this runs; you can ignore them.

```python
xs = cross_section(fl, lats, lons)
xs
```

## Plot the cross-section

The surface pressure is in pascal (Pa), so we first convert it to hPa to match the pressure axis of the plot. Then we draw the temperature, fill the area below the surface pressure in black to show the ground, and label the axes.

```python
# convert the surface pressure from Pa to hPa
xs["sp"] = xs["sp"] / 100
xs["sp"].attrs["units"] = "hPa"
```

```python
def fmt_lat(lat):
    lat = int(round(lat))
    return f"{abs(lat)}°N" if lat > 0 else f"{abs(lat)}°S" if lat < 0 else "0°"


def fmt_lon(lon):
    lon = int(round(lon))
    return f"{abs(lon)}°E" if lon > 0 else f"{abs(lon)}°W" if lon < 0 else "0°"


# 7 evenly spaced points along the line
idx = np.linspace(0, len(lats) - 1, 7).round().astype(int)
tick_labels = [f"{fmt_lat(lats[i])}\n{fmt_lon(lons[i])}" for i in idx]

chart = ekp.Subplot(figsize=(10, 5))
chart.contourf(xs["t"], x="distance", y="level")
chart.figure.legend()
# orography: black area from the surface pressure down to 1000 hPa
chart.fill_between(xs["sp"], 1000, x=xs["distance"], color="black", alpha=1)

ax = chart.ax
ax.set_ylim(1000, 300)  # 1000 hPa at the bottom, 300 hPa at the top
ax.set_xticks(xs["distance"].values[idx])
ax.set_xticklabels(tick_labels)
ax.set_xlabel("Position along the cross-section")
ax.set_ylabel("Pressure (hPa)")
ax.set_title(
    f"Temperature (K) and orography (black) – {base_date:%Y-%m-%d %H} UTC, step {step.total_seconds() / 3600:g} h"
)

chart.show()
```
